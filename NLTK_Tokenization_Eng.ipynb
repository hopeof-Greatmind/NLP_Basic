```python
!pip install konlpy
from konlpy.tag import Okt
okt = Okt()
```

```python
text = "자연어 실습은 재미있습니다!"

#형태소 단위 분리
print(okt.morphs(text))

#품사 태깅
print(okt.pos(text))

#명사 추출
print(okt.nouns(text))
```
